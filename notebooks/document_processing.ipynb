# Day 1 — Document Processing Exploration

This notebook explores the ingestion pipeline in `src/ingestion.py`:
loading the policy `.txt` files, cleaning them, tagging metadata, and
chunking. See `chunking_notes.md` in this folder for the chunk-size
experiment writeup.

```python
import sys
sys.path.append('../src')
from ingestion import load_documents, clean_document, chunk_document, print_document_summary, process_all

documents = load_documents()
print_document_summary(documents)
```

```python
# Inspect cleaning on one document
raw = documents['travel_policy_india.txt']
cleaned = clean_document(raw)
print(cleaned[:500])
```

```python
# Inspect chunking on one document
chunks = chunk_document('travel_policy_india.txt', cleaned)
for c in chunks:
    print(c.chunk_id, '-', c.char_count, 'chars')
    print(c.text[:150], '...\n')
```
